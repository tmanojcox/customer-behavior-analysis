```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/customer_shopping_behavior.csv')
df.head()
```

```python
df["Review Rating"]=df["Review Rating"].fillna(
    df.groupby("Category")["Review Rating"].transform('median')
)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

```python
df=df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

```python
labels= ['younger_adult','adult','middle_age','senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head()
```

```python
# create column purchase_frequency_days
frequency_mapping={

    'Annually' :         365,
'Bi-Weekly'    :     14,
'Every 3 Months':    90,
'Fortnightly'    :   14,
'Monthly'         :  30,
'Quarterly'        : 90,
'Weekly'       :7
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']].head()
```

```python
df=df.drop(columns=['promo_code_used'],axis=1)
```

```python
df.head()
```

```python
df.to_csv("customer_data_cleaned.csv", index=False)
```
